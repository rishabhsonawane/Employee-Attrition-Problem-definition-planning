```python
numerical_features = df.select_dtypes(include=['int64', 'float64']).columns.tolist()
categorical_features = df.select_dtypes(include=['object']).columns.tolist()

print("Numerical Features:")
print(numerical_features)

print("\nCategorical Features:")
print(categorical_features)
```

```python
missing_values = df.isnull().sum()

missing_table = pd.DataFrame({
    'Column': missing_values.index,
    'Missing Values': missing_values.values,
    'Missing Percentage': (missing_values.values / len(df)) * 100
})

missing_table = missing_table.sort_values(
    by='Missing Values',
    ascending=False
)

missing_table
```

```python
missing = df.isnull().sum()

missing = missing[missing > 0]

if len(missing) == 0:
    print("No missing values found in the dataset.")
else:
    plt.figure(figsize=(10, 5))
    missing.plot(kind='bar')
    plt.title("Missing Values by Feature")
    plt.xlabel("Features")
    plt.ylabel("Number of Missing Values")
    plt.xticks(rotation=45)
    plt.show()
```

```python
df_corr = df.copy()

df_corr['Attrition_Encoded'] = df_corr['Attrition'].map({
    'Yes': 1,
    'No': 0
})

print(df_corr[['Attrition', 'Attrition_Encoded']].head())
```

```python
correlation_matrix = df_corr.select_dtypes(
    include=['int64', 'float64']
).corr()

correlation_matrix
```

```python
plt.figure(figsize=(18, 14))

sns.heatmap(
    correlation_matrix,
    cmap='coolwarm',
    center=0,
    linewidths=0.5
)

plt.title("Correlation Matrix of Numerical Features")

plt.tight_layout()
plt.show()
```

```python
attrition_correlation = (
    correlation_matrix['Attrition_Encoded']
    .drop('Attrition_Encoded')
    .sort_values()
)

print("Features negatively correlated with Attrition:")
print(attrition_correlation.head(10))

print("\nFeatures positively correlated with Attrition:")
print(attrition_correlation.tail(10))
```

```python
top_correlations = (
    attrition_correlation
    .abs()
    .sort_values(ascending=False)
    .head(10)
)

print("Top 10 features by absolute correlation with Attrition:")
print(top_correlations)
```
